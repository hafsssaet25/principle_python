```
num_1 = int(input("Entre the first number:  "))
num_2 = int(input("Entre the second number: "))
operador = input("choose the operator (+,-,*,/): ")
if operador == "+" :
    print(num_1 + num_2)

elif operador == "*":
    print(num_1 * num_2)

elif operador == "/" :
  if num_2 == 0 :
    print("You cannot divide by 0 ")
  else :
    print(num_1 / num_2)

elif operador == "-":
  print(num_1 - num_2)

else :
  print("Error")





```
